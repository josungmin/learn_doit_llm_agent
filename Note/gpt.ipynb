````python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)

models = client.models.list()
````

**rich.pretty.pprint**

`rich.pretty.pprint`는 파이썬 객체를 사람이 읽기 좋은 형태로 출력하는 함수입니다. 기본 `print`보다 중첩된 리스트와 딕셔너리의 구조를 보기 쉽게 정렬하고, 긴 값은 줄바꿈하여 보여줍니다. 터미널에서는 자료형과 문자열 등을 색상으로 구분해 디버깅할 때 유용합니다.

`rich`가 설치되어 있지 않다면 프로젝트 폴더에서 `uv add rich` 명령으로 설치합니다.

````python
from rich.pretty import pprint
# created기준으로 내림차순 정렬 후 상위 15개 모델만 가져오기
models = sorted(models, key=lambda model: model.created, reverse=True)[:15]
pprint(models)
````

# GPT-5.6 Luna와 `temperature`

## 1. 왜 `temperature=0.1`이 동작하지 않는가?

`gpt-5.6-luna`에서는 `temperature`를 사용자가 조절할 수 없고 기본값 `1`만 허용됩니다.

```python
client.chat.completions.create(
    model="gpt-5.6-luna",
    temperature=0.1,
)
```

위 코드는 다음과 같은 오류를 반환합니다.

```text
Unsupported value: 'temperature'
does not support 0.1 with this model.
Only the default (1) value is supported.
```

이는 Python SDK 문제가 아니라 **모델 자체의 파라미터 지원 제한**입니다.

즉,

```text
API에 temperature 필드가 존재한다
≠
모든 모델이 temperature 조절을 지원한다
```

최신 reasoning 모델은 전통적인 sampling 제어보다 reasoning 수준과 출력 방식 제어를 중심으로 설계되어 있습니다.

---

## 2. GPT-5.6 Luna에서 중요한 파라미터

GPT-5.6 계열에서는 `temperature`보다 다음 설정이 더 중요합니다.

| 파라미터 | 역할 |
|---|---|
| `reasoning.effort` | 모델이 reasoning에 사용할 계산량 조절 |
| `reasoning.mode` | reasoning 실행 방식 조절 |
| `reasoning.context` | 이전 reasoning 정보 활용 방식 |
| `verbosity` | 최종 답변의 상세도 |
| `max_output_tokens` | 최대 출력 길이 |
| `tools` | 사용할 tool 정의 |
| `tool_choice` | tool 선택 방식 |
| Structured Outputs | JSON Schema 등으로 출력 구조 제한 |
| streaming | 스트리밍 응답 |

예:

```python
response = client.responses.create(
    model="gpt-5.6-luna",
    input="이 문제를 설명해줘.",
    reasoning={
        "effort": "low"
    },
)
```

개념적으로는 다음처럼 이해할 수 있습니다.

```text
기존 모델

temperature
   ↓
얼마나 랜덤하게 생성할 것인가?


GPT-5.6

reasoning.effort
   ↓
얼마나 많이 생각할 것인가?
```

---

## 3. `verbosity`와 `temperature`의 차이

두 파라미터는 목적이 다릅니다.

```text
temperature
    ↓
token 선택의 확률 분포를 변화시킴

verbosity
    ↓
최종 답변을 얼마나 자세하게 작성할지 조절
```

예:

```python
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[...],
    verbosity="low",
)
```

따라서 `verbosity="low"`는 `temperature=0.1`의 대체재가 아닙니다.

---

## 4. `temperature`를 사용할 수 있는 다른 모델

전통적인 non-reasoning 모델에서는 `temperature`를 직접 조절하는 방식이 일반적입니다.

예:

```python
client.chat.completions.create(
    model="gpt-4.1",
    temperature=0.1,
    messages=[...],
)
```

대략 다음과 같이 구분할 수 있습니다.

```text
전통적인 GPT 모델
    ↓
temperature 중심의 sampling 제어

최신 GPT-5.6 reasoning 모델
    ↓
reasoning.effort
verbosity
tools
structured output
```

단, API에 `temperature` 파라미터가 존재하더라도 실제 지원 여부는 모델마다 다를 수 있으므로 모델 문서를 확인해야 합니다.

---

## 5. GPT-5.6 Luna를 결정론적으로 사용할 수 있는가?

엄밀한 의미에서는 **완전한 결정론을 보장할 수 없습니다.**

원하는 조건이 다음이라면:

```text
동일 prompt
+
동일 parameters
=
항상 완전히 동일한 response
```

GPT-5.6 Luna에서는 이를 보장하는 방법이 없습니다.

특히 다음과 같은 전통적인 설정을 사용할 수 없습니다.

```python
temperature=0
```

따라서 greedy decoding 방식으로 모델을 강제할 수도 없습니다.

---

## 6. 결과를 최대한 일관되게 만드는 방법

완전한 결정론 대신 **출력의 자유도를 줄이는 방식**이 현실적입니다.

예를 들어 자유 텍스트 대신 가능한 출력을 제한합니다.

```python
response = client.responses.create(
    model="gpt-5.6-luna",
    input="""
다음 문장을 분류하라.

반드시 다음 중 하나만 반환:
- positive
- negative
- neutral

문장:
제품은 좋은데 배송이 느렸다.
""",
    reasoning={
        "effort": "none"
    },
)
```

더 안정적인 방법은 Structured Outputs를 사용하는 것입니다.

```text
자유 텍스트
    ↓
가능한 출력이 많음
    ↓
결과 변동 가능성 증가


JSON Schema / enum
    ↓
출력 가능한 형태 제한
    ↓
결과 변동 가능성 감소
```

실무에서는 다음 조합이 중요합니다.

```text
명확한 prompt
+
Structured Outputs
+
enum / JSON Schema
+
고정된 tool schema
+
낮고 일관된 reasoning 설정
```

---

## 정리

| 질문 | 답 |
|---|---|
| 왜 `temperature=0.1`이 안 되는가? | Luna 모델이 기본값 `1`만 지원하기 때문 |
| SDK 문제인가? | 아니며 모델 capability 제한 |
| `temperature=0` 사용 가능? | 불가능 |
| Luna의 주요 제어값 | `reasoning.effort`, `reasoning.mode`, `verbosity`, tools 등 |
| `verbosity`가 temperature 대체인가? | 아님 |
| 다른 모델에서는 temperature 사용 가능? | 일부 non-reasoning 모델에서 가능 |
| Luna를 완전히 deterministic하게 사용 가능? | 보장 불가 |
| 일관성을 높이는 방법 | 명확한 prompt + Structured Outputs + 제한된 출력 공간 |

핵심적으로 GPT-5.6 Luna는 **sampling parameter를 직접 미세 조정하는 모델이라기보다, reasoning 동작과 출력 형식을 높은 수준에서 제어하는 모델**로 이해하는 것이 좋습니다.
