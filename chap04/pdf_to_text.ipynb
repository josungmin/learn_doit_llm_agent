PDF 문서에서 노이즈가 되는 머리글과 바닥글을 제외하고 순수한 본문 텍스트만 추출하는 데이터 전처리 스크립트입니다.

```python
import pymupdf
import os

pdf_file_path = "data/과정기반 작물모형을 이용한 웹 기반 밀 재배관리 의사결정 지원시스템 설계 및 구축.pdf"
doc = pymupdf.open(pdf_file_path)
```

1. import pymupdf: PDF 문서를 다루기 위한 강력한 라이브러리입니다. (https://pypi.org/project/pymupdf/)
2. import os: 파일 경로 추출 등 운영체제와 상호작용하기 위해 사용합니다.
3. 지정된 경로의 PDF 파일을 열어 doc 객체에 저장합니다.

```python
header_height = 80
footer_height = 80

full_text = ''

for page in doc:
    rect = page.rect
    
    header = page.get_text(clip=(0, 0, rect.width , header_height))
    footer = page.get_text(clip=(0, rect.height - footer_height, rect.width , rect.height))
    text = page.get_text(clip=(0, header_height, rect.width , rect.height - footer_height))

    full_text += text + '\n------------------------------------\n'
```

1. 위쪽 80px을 머리글 영역으로, 아래쪽 80px을 바닥글 영역으로 정의합니다.
2. 문서의 모든 페이지를 순회하면서(for page in doc:) 각 페이지의 전체 크기(rect)를 가져옵니다.
3. clip=(x0, y0, x1, y1) 옵션을 사용하여 페이지의 특정 사각형 영역에서만 텍스트를 추출합니다.
    - header: 페이지 최상단부터 header_height(80)까지의 영역.
    - footer: 페이지 최하단에서 footer_height(80)만큼 위로 올라간 곳부터 최하단까지의 영역.
    - text: 머리글 아래부터 바닥글 위까지, 즉 순수 본문 영역입니다.
4. 추출한 본문 텍스트(text)만 full_text 변수에 누적해서 더합니다. 각 페이지가 끝날 때마다 구분선(---)을 추가하여 가독성을 높였습니다.

```python
pdf_file_name = os.path.basename(pdf_file_path)
pdf_file_name = os.path.splitext(pdf_file_name)[0] # 확장자 제거

txt_file_path = f'output/{pdf_file_name}_with_preprocessing.txt'

with open(txt_file_path, 'w', encoding='utf-8') as f:
    f.write(full_text)
```

1. os.path.basename을 통해 전체 경로에서 파일명(과정기반...pdf)만 분리합니다.
2. os.path.splitext를 통해 .pdf 확장자를 떼어내고 순수 파일 이름만 pdf_file_name에 저장합니다.
3. 출력될 파일의 경로를 chap04/output/파일명_with_preprocessing.txt 형식으로 생성합니다.
4. 성된 경로에 UTF-8 인코딩으로 파일을 열어 누적된 full_text를 저장합니다.
