# Pydantic의 BaseModel로 출력 형태 정의하기

## 파이단틱 (Pydantic)

**파이썬의 타입 힌트를 기반으로 런타임에 데이터의 유효성을 엄격하게 검증하고 자동으로 변환해 주는 데이터 검증 라이브러리입니다.**

* **런타임 검증 강제:** 기존 파이썬 환경에서는 무시되던 타입 힌트를 실제 프로그램 실행 시점의 강력한 유효성 검증 규칙으로 승격시킵니다.
* **데이터 차단 및 예외 발생:** 명시된 타입으로 처리할 수 없는 잘못된 데이터가 입력되면, 즉시 `ValidationError`를 발생시켜 비정상적인 데이터 유입을 차단합니다.
* **자동 형변환 (Type Casting):** 데이터 타입이 정확히 일치하지 않더라도 논리적으로 변환이 가능한 형태(예: 정수형이 필요한 곳에 문자열 `"123"` 입력)라면, 자동으로 요구되는 타입(정수 `123`)으로 변환하여 처리합니다.

---

### 1. 파이썬의 동적 타이핑 (Dynamic Typing)과 도입 배경

**파이썬은 실행 중에 데이터 타입이 결정되고 유연하게 변할 수 있어 예기치 않은 데이터 오류가 발생하기 쉽기 때문에 파이단틱과 같은 검증 도구가 필요합니다.**

* **실행 시점 타입 결정:** 프로그램이 실행되는 런타임(Runtime)에 데이터의 타입이 결정되며, 실행 중에도 데이터 타입이 변경될 수 있습니다.
* **객체 참조 방식:** 변수는 특정 데이터 타입에 고정되지 않고, 메모리에 할당된 데이터(객체)를 참조(Reference)하는 역할만 수행합니다.
* **런타임 오류 위험:** 제약이 없는 유연성 때문에 잘못된 타입의 데이터가 유입되어도 코드가 실행되기 전까지는 차단하기 어려워 치명적인 런타임 에러로 이어질 수 있습니다.

---

### 2. 타입 힌트 (Type Hints)와 파이단틱의 연계

**파이썬의 타입 힌트는 본래 실행 시 무시되는 기능이지만, 파이단틱은 이를 활용하여 실제 데이터 동작을 통제하는 명확한 기준으로 사용합니다.**

* **타입 명시:** 변수나 함수에 요구되는 데이터 타입을 코드 상에 명시적으로 작성하는 기능입니다.
* **런타임 강제성 없음:** 파이썬은 프로그램 실행 시 이 힌트를 강제하지 않으므로, 기본 환경에서는 명시된 타입과 다른 데이터가 입력되어도 에러 없이 그대로 통과시킵니다.
* **정적 분석 및 확장:** 본래는 정적 분석(Static Analysis) 도구를 통한 코드 오류 사전 검출 및 개발자 간의 명세서 목적으로 쓰이나, 파이단틱은 이 힌트를 읽어들여 런타임 데이터 유효성 검증의 핵심 지표로 활용합니다.

```python
from pydantic import BaseModel, Field
from typing import Literal

class Task(BaseModel):
    agent: Literal[
        "content_strategist",
        "communicator",
        "web_search_agent",
        "vector_search_agent",
    ] = Field(
        ...,
        description="""
        작업을 수행하는 agent의 종류.
        - content_strategist: 콘텐츠 전략을 수립하는 작업을 수행한다. 사용자의 요구사항이 명확해졌을 때 사용한다. AI 팀의 콘텐츠 전략을 결정하고, 전체 책의 목차(outline)를 작성한다. 
        - communicator: AI 팀에서 해야 할 일을 스스로 판단할 수 없을 때 사용한다. 사용자에게 진행상황을 사용자에게 보고하고, 다음 지시를 물어본다.
        - web_search_agent: 웹 검색을 통해 목차(outline) 작성에 필요한 정보를 확보한다.
        - vector_search_agent: 벡터 DB 검색을 통해 목차(outline) 작성에 필요한 정보를 확보한다.
        """
    )
	
    done: bool = Field(
        ..., 
        description="종료 여부"
    )

    description: str = Field(
        ..., 
        description="어떤 작업을 해야 하는지에 대한 설명"
    )
    
    done_at: str = Field(
        ..., 
        description="할 일이 완료된 날짜와 시간"
    )
	
    def to_dict(self):
        return {
            "agent": self.agent,
            "done": self.done,
            "description": self.description,
            "done_at": self.done_at
        }  
```
