# 프롬프트 엔지니어링이란 무엇인가요?

프롬프트 엔지니어링은 AI 모델, 특히 LLM이 원하는 응답을 생성할 수 있도록 프롬프트를 설계하고 최적화하는 기술입니다. 프롬프트를 신중하게 작성하면 모델이 사용자의 의도를 이해하고 의미 있는 방식으로 응답하는 데 도움이 되는 컨텍스트, 안내, 예시를 모델에 제공할 수 있습니다. AI에 로드맵을 제공하여 내가 염두에 두고 있는 특정 결과물을 얻도록 방향을 조정한다고 생각하면 됩니다.

https://cloud.google.com/discover/what-is-prompt-engineering?hl=ko#what-is-prompt-engineering

### AI를 위한 프롬프트란 무엇인가요?
AI 컨텍스트에서 프롬프트는 구체적인 응답을 도출하기 위해 모델에 제공하는 입력입니다. 간단한 질문이나 키워드부터 복잡한 안내, 코드 스니펫, 창의적인 글쓰기 샘플에 이르기까지 다양한 형태를 취할 수 있습니다. 프롬프트의 효과는 AI 출력의 품질과 관련성에 직접적인 영향을 미칩니다.

## 프롬프트 엔지니어링 1 - 페르소나(Persona)

AI 모델에게 특정 역할, 직업, 성격 또는 정체성을 부여하여 답변의 질을 높이는 기법입니다.
- **목적**: 모델이 특정 관점이나 전문성을 띠고 답변하도록 유도합니다.
- **효과**: 답변의 톤앤매너(어조), 어휘 선택, 전문성 수준이 페르소나에 맞춰 조정되므로, 사용자가 원하는 맞춤형 결과물을 얻기 용이해집니다.
- **예시**: "너는 10년 차 시니어 소프트웨어 엔지니어 구루야.", "초등학생에게 설명하듯 쉽고 친절하게 알려줘."

##### chap03/sec01/p01/joker_in_batman.py

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
from rich.pretty import pprint

load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)

response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role": "system", "content": "너는 배트맨에 나오는 조커야. 조커의 악당 캐릭터에 맞게 답변해줘"},
        {"role": "user", "content": "세상에서 누가 제일 아름답니?"},
    ]
)

pprint(response)

print('----')
print(response.choices[0].message.content) 
```

##### chap03/sec01/p01/mirror_in_snow_white.py

```python
response = client.chat.completions.create(
  model="gpt-5.6-luna",
  messages=[
    {"role": "system", "content": "너는 백설공주 이야기 속의 거울이야. 그 이야기 속의 마법 거울의 캐릭터에 부합하게 답변해줘."},
    {"role": "user", "content": "세상에서 누가 제일 아름답니?"},
  ]
)

pprint(response)

print('----')
print(response.choices[0].message.content) 
```

## 프롬프트 엔지니어링 2 - 예시를 제시하는 프롬프팅
프롬프트 엔지니어링에서는 AI 모델에게 제공하는 예시(Shot)의 개수에 따라 프롬프팅 기법을 크게 제로샷, 원샷, 퓨샷으로 구분합니다.

### 1) 제로샷 프롬프팅 (Zero-shot Prompting)
- 예시를 전혀 제공하지 않고, 오직 지시사항(명령)만으로 작업을 수행하도록 하는 방식입니다.
- 직관적이고 난이도가 낮은 작업에 적합 (ex. 단순한 번역, 요약, 일반 상식 질문 등)
- 엄격한 출력 형식이 요구되지 않으며, 모델이 학습한 기본 능력만으로 충분히 해결할 수 있을 때 주로 사용합니다.

##### chap03/sec01/p02/no_prompting.py

```python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    # 유치원생 유치원생 페르소나의 '엉뚱함'과 '다양성' 유도하기 위하여 temperature 값을 1에 가깝게 설정한다.
    temperature=0.9,
    messages=[
        {"role": "system", "content": "너는 유치원 학생이야. 유치원생처럼 답변해줘."},
        {"role": "user", "content": "오리"},
    ]
)

pprint(response)

print('----')
print(response.choices[0].message.content) 
```

### 2) 원샷 프롬프팅 (One-shot Prompting)
- 단 하나의 예시(입력과 출력의 쌍)를 모델에게 제공하는 방식입니다.
- 특정한 출력 형식이나 구조를 명확히 강제해야 할 때 적합 (ex. Json 등)
- 제로샷으로 지시했을 때 모델이 불필요한 사족을 덧붙이거나 지정된 형식을 벗어나는 문제를 효과적으로 교정합니다.

##### chap03/sec01/p02/one_shot.py

```python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    # 유치원생 유치원생 페르소나의 '엉뚱함'과 '다양성' 유도하기 위하여 temperature 값을 1에 가깝게 설정한다.
    # 단, 정해진 패턴(동물 -> 의성어)을 정확하게 출력하는것이 목적이라면 temperature를 낮게 설정해야하는것 아닌가?
    temperature=0.9,
    messages=[
        {"role": "system", "content": "너는 유치원 학생이야. 유치원생처럼 답변해줘."},
        {"role": "user", "content": "참새"},
        {"role": "assistant", "content": "짹짹"},
        {"role": "user", "content": "오리"},
    ]
)

pprint(response)

print('----')
print(response.choices[0].message.content) 
```

### 3) 퓨샷 프롬프팅 (Few-shot Prompting)
- 여러 개(보통 2~5개 이상)의 다양한 예시를 모델에게 제공하는 방식입니다.
- 복잡한 데이터 추출, 논리적 추론 문제, 특정 도메인에 특화된 텍스트 생성 등 고난도 작업에 적합
- 다양한 엣지 케이스(예외 상황)에 대한 처리 기준이나 일관된 패턴을 모델에게 정확히 안내하고자 할 때 필수적입니다.

##### chap03/sec01/p02/few_shot.py

```python
response = client.chat.completions.create(
    model="gpt-5.4-mini",
    # 유치원생 유치원생 페르소나의 '엉뚱함'과 '다양성' 유도하기 위하여 temperature 값을 1에 가깝게 설정한다.
    # 단, 정해진 패턴(동물 -> 의성어)을 정확하게 출력하는것이 목적이라면 temperature를 낮게 설정해야하는것 아닌가?
    temperature=0.9,
    messages=[
        {"role": "system", "content": "너는 유치원 학생이야. 유치원생처럼 답변해줘."},
        {"role": "user", "content": "참새"},
        {"role": "assistant", "content": "짹짹"},
        {"role": "user", "content": "말"},
        {"role": "assistant", "content": "히이잉"},
        {"role": "user", "content": "개구리"},
        {"role": "assistant", "content": "개굴개굴"},
        {"role": "user", "content": "뱀"},
    ]
)

pprint(response)

print('----')
print(response.choices[0].message.content) 
```
